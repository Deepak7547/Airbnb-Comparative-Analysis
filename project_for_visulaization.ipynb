#**Project Name** : Airbnb Market Analysis Dashboard.





##**Project Summary**

This project analyzes Airbnb listings from two major U.S. cities, Chicago and New Orleans. The objective is to compare property availability, pricing trends, neighbourhood popularity, room types, host behavior, and customer reviews. Data cleaning and transformation were performed using Python (Pandas) in Google Colab, and an interactive dashboard was developed in Power BI to visualize key insights and comparisons between the two cities.

##Project Objectives

- Analyze Airbnb listings across Chicago and New Orleans.

- Identify the most popular neighbourhoods based on listing counts.

- Compare average property prices across locations and room types.

- Understand the distribution of different room types.

- Analyze the relationship between pricing and customer reviews.

- Study host behavior and listing patterns.

- Build an interactive Power BI dashboard for business insights and decision-making.

# Power BI Dashboard

The dashboard was developed in Power BI and consists of:

1. Overview of Airbnb
2. Property Analysis
3. Pricing Analysis
4. Host Analysis

Page 1 – Overview of Airbnb

Page1.png

**Why these visuals were used?**

- KPI Cards were used to quickly display total listings, average price, total reviews, and total hosts.

- Donut Chart was used to show room type distribution.
Slicer was added for city-wise comparison.

**Key Insights**

- Chicago has more Airbnb listings than New Orleans.

- Entire Home/Apt is the most common room type in both cities.

- Average pricing differs between the two cities.

Page 2 – Property Analysis

page2.png

**Why these visuals were used?**

- Bar Charts were used to compare neighbourhood popularity and pricing.

- Donut/Column Chart was used to visualize room type distribution.

**Key Insights**

- Certain neighbourhoods dominate Airbnb listings.

- Popular neighbourhoods generally have higher listing counts.

- Entire Home/Apt properties represent the largest share of listings.

Page 3 – Pricing Analysis

page3.png

**Why these visuals were used?**

- Bar Chart was used to compare average prices by room type.

- Donut Chart was used to show price category distribution.

- Scatter Plot was used to identify the relationship between price and reviews.

- Column Chart was used for city-level price comparison.

**Key Insights**

- Hotel rooms have the highest average prices.

- Most listings fall under the Standard and Budget price categories.

- Higher prices do not always result in more reviews.

- Chicago and New Orleans show different pricing patterns.

Page 4 – Host Analysis

page4.png

**Why these visuals were used?**

- Bar Chart highlights top hosts by listing count.

- Scatter Plots help identify relationships between host activity, reviews, and pricing.

**Key Insights**

- A small number of hosts manage a large number of listings.

- Hosts with more listings do not always receive more reviews.

- Pricing varies significantly among hosts with similar listing counts.

- Host behaviour differs across cities.
